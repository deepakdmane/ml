# import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import csv
import random

# Define the number of records to generate
num_records = 100  # You can change this to the desired number

# Define lists of states and a range for numerical values
states = ["California", "Texas", "New York", "Florida", "Illinois"]
profit_range = (10000, 100000)  # Example range, adjust as needed
rnd_range = (5000, 50000)
admin_range = (10000, 30000)
marketing_range = (5000, 30000)

# Create the data
data = [["R&D Spend", "Administration", "Marketing Spend", "State","Profit"]]
for _ in range(num_records):
    profit = random.randint(*profit_range)
    rnd = random.randint(*rnd_range)
    admin = random.randint(*admin_range)
    marketing = random.randint(*marketing_range)
    state = random.choice(states)
    data.append([rnd, admin, marketing, state,profit])

# Save the data to a CSV file
file_path = "company_data1.csv"
with open(file_path, "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print(f"CSV file '{file_path}' created successfully.")
```

# Import DataSET

```python
try:
    dataset = pd.read_csv('company_data1.csv')
    display(dataset.head())
except FileNotFoundError:
    print("Error: 'Data.csv' not found. Please ensure the file exists in the current directory.")
except pd.errors.EmptyDataError:
    print("Error: 'Data.csv' is empty.")
except pd.errors.ParserError:
    print("Error: 'Data.csv' could not be parsed. Please check the file format.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

x = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
print(x)
```

```python
print(y)
```

# Encode Categorical Data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])], remainder='passthrough')
X = np.array(ct.fit_transform(x))
```

```python
print(X)
```

# Splitting data into training data and test data

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.01, random_state = 0)
```

```python
X_train, X_test, y_train, y_test
```

#Training the multiple linear Regression model on the training set

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(X_train, y_train)
```

#Predicting the test set results

```python
ypred = regression.predict(X_test)
np.set_printoptions(precision=2)
print(np.concatenate((ypred.reshape(len(ypred),1), y_test.reshape(len(y_test),1)),1))
```
